## 2. The question: decision, action, cost of a wrong call

**Research question:** Which pages should be prioritized for human review as potential content-refresh opportunities based on their observed search-performance and content signals?

The decision is which pages should be reviewed first. A content or SEO team would act on this ranking by reviewing the highest-priority pages and deciding whether an update, improvement, or further investigation is appropriate.

A false positive could cause the team to spend limited time reviewing a page that does not need attention. A false negative could cause a page with a genuine opportunity or decline to be overlooked. Therefore, the goal is to support prioritization and human review rather than make an automatic decision.

## 3. Quick look at the data (2-3 real numbers)

The starter dataset provides page-level observations that are relevant to the content-refresh question. I will use the number of pages, median search impressions over 90 days, and median days since the last update as an initial check of the dataset.

These measurements show whether there is enough variation in page visibility and content freshness to make page-level prioritization worth investigating over the next seven weeks. The numbers below are calculated directly from the starter CSV.

```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.
# Load the starter dataset and show 2-3 real numbers

import os
import pandas as pd

REPO_URL = "https://github.com/Uman-66/Flyrank-Internship"
REPO_DIR = "Flyrank-Internship"

# Clone the repository if it is not already available in the Colab runtime
if not os.path.isdir(REPO_DIR):
    import subprocess
    subprocess.run(
        ["git", "clone", "--depth", "1", REPO_URL, REPO_DIR],
        check=True
    )

# Move into the repository
os.chdir(REPO_DIR)

# Load the starter CSV
DATA_PATH = "data/raw/content_refresh_anonymized.csv"
df = pd.read_csv(DATA_PATH)

# Calculate real numbers from the dataset
number_of_pages = len(df)
median_impressions = df["impressions_90d"].median()
median_days_since_update = df["days_since_last_update"].median()

print(f"Number of pages: {number_of_pages:,}")
print(f"Median impressions over 90 days: {median_impressions:,.1f}")
print(f"Median days since last update: {median_days_since_update:,.1f}")
```

This work can identify **observed and measured patterns** in the available page-level data and provide **directional, decision-support** rankings for pages that may deserve human review.

I cannot claim that the analysis proves that one factor causes changes in search performance. An observed relationship between content age, impressions, CTR, or another signal and page performance would not by itself establish causation.

The results should therefore be treated as evidence for prioritization and decision support rather than causal proof.
